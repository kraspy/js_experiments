JavaScript - язык со слабой динамической типизацией

###  `Array => String`

```typescript
console.log([].toString())         // ''
console.log([1, 2, 3].toString())  // '1,2,3'
```

### `Number`
Объект берет значение из `valueOf()`, а если не получается, то из `toString()`

```typescript
let numbers = [
    Number(true),       // 1
    Number(false),      // 0
    Number(undefined),  // NaN
    Number(null),       // 0
    Number([]),         // 0
    Number([1, 2]),     // NaN
    Number('Z'),        // NaN
    Number('1'),        // 1
    Number(''),         // 0
]

numbers
```

### `Boolean`

#### `Ложные` значения:

```typescript
let falsy = [
    Boolean(undefined), // false
    Boolean(null),      // false
    Boolean(false),     // false
    Boolean(NaN),       // false
    Boolean(0),         // false
    Boolean(''),        // false
]

falsy
```

#### `Истинные` значения:
Любые значения, которые не входят в список ложных

```typescript
let true_bools = [
    Boolean([]),   // true
    Boolean(' '),  // true
    Boolean({}),   // true
]

true_bools
```

### `==` и `===`
- `==`  Допускает преобразование типов
- `===` Не допускает преобразование типов
- `NaN` Никогда не равен самому себе
- +0 Всегда равен -0
- `==` и `===` ведут себя одинаково при сравнении объектов
- `!=` и `!==` вычисляют сначала равенство, а потом отрицают результат

### `null` и `undefined`

```typescript
let primitives = [
    null == undefined,       // true
    undefined == null,       // true
    undefined == undefined,  // true
    null == null,            // true
    null >= 0,               // true
]

primitives
```

### `Объекты` и `примитивы`
При сравнении объектов с примитивами, у объектв вызываются `toValue()` или `toString()`

```typescript
let a = 'abc';
let b = Object(a);

console.log(a == b);   // true,  т.к. "abc" == "abc"
console.log(a === b);  // false, т.к. "abc" != Object("abc")
```

❗ `Но` не в случае с `null` или `undefined`\
`null` и `undefined` не умеют упаковываться,\
поэтому Object(null) или Object(undefined), создают обычные пустые объекты, типа Object()

```typescript
let vals = [];

// null
let n = null;
let on = Object(n);
vals.push(n == on);      // false, т.к. null != Object();

// undefined
let u = undefined;
let ou = Object(u);
vals.push(u == ou);      // false, т.к. undefined != Object();

// NaN
let nan = NaN;
let onan = Object(a);
vals.push(nan == onan);  // false, т.к. NaN != new Number(a);


console.log(vals);


NaN != NaN;              // true
```

### Сравнение ложных значений

```typescript
let falsy_vals = [
    '0' == null,         // false
    '0' == undefined,    // false
    '0' == false,        // true
    '0' == NaN,          // false
    '0' == 0,            // true
    '0' == '',           // false
    false == null,       // false
    false == undefined,  // false
    false == NaN,        // false
    false == 0,          // true
    false == '',         // true
    false == [],         // true
    false == {},         // false
    '' == null,          // false
    '' == undefined,     // false
    '' == NaN,           // false
    '' == 0,             // true
    '' == [],            // true
    '' == {},            // false
    0 == null,           // false
    0 == undefined,      // false
    0 == NaN,            // false
    0 == [],             // true
    0 == {},             // false
]

falsy_vals
```
