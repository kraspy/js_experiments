## use strict

В модулях use strict включен по умолчанию.

### Значение контекста `this`
`this` принимает одно из двух значений:
- `undefined` — в строгом режиме,
- `window` — в обычном.

```typescript
function globalFunction() {
  'use strict';
  console.log(this);
}
globalFunction();
```

### Необъявленные переменные
Если переменная не объявлена, в неё нельзя записать значение:

```typescript
greeting = 'Привет';
```

### Изменение аргументов функции
В обычном режиме можно менять значения аргументов, обращаясь к ним по индексам.

```typescript
function consoleDog(dog) {
  arguments[0] = 'Чихуахуа'; // Перезаписываем аргумент
  console.log(dog);
}
consoleDog('Джек Рассел'); // "Чихуахуа"
```

```typescript
'use strict';
function consoleDog(dog) {
  arguments[0] = 'Чихуахуа'; // Попытка изменить
  console.log(dog);
}
consoleDog('Джек Рассел'); // "Джек Рассел"
```

### Уникальность имен параметров
В строгом режиме параметры функции не могут иметь одинаковые имена

```typescript
'use strict';
function consoleDog(dog, dog) { // SyntaxError
  console.log(dog);
}
```
